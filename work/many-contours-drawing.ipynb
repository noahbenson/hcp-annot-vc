```python
import neuropythy as ny
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Get the HCP-lines dataset.
dset = ny.data['hcp_lines']

# To use individual subject contours (not aligned), we can use native_path_traces:
#traces = dset.native_path_traces

# For aligned traces, we used fsaverage:
traces = dset.fsaverage_path_traces
```

```python
# Traces is a nested dictionary:
# for example, traces['A1'][111312]['lh']['iso_angle']['V1_ventral']
# is the PathTrace object for:
#  * rater A1
#  * subject 111312
#  * hemisphere LH
#  * V1_ventral contour (eccen contours are in 'iso_eccen')
#
# To get points from a PathTrace:
#  (x, y) = pathtrace.points
# For example:
#  (x, y) = traces['A1'][111312]['lh']['iso_angle']['V1_ventral'].points
```

```python
# Parameters of what to plot:
rater = 'mean'
contour_names = {
    'V1_ventral': 'r',
    'V1_dorsal': 'r',
    'V2_ventral': 'c',
    'V2_dorsal': 'c',
    'V3_ventral': '0.5',
    'V3_dorsal': '0.5',
    '7': 'k'}
# Parameters of the figure:
figsize = (8,4)
dpi = 288
alpha = 0.1
lw = 0.25

# Make the figure first:
(fig, axs) = plt.subplots(1, 2, figsize=figsize, dpi=dpi)

for (h, ax) in zip(['lh', 'rh'], axs):
    # Plot fsaverage map?
    for sid in dset.subject_list:
        try:
            sub_traces = traces[rater][sid][h]
        except Exception:
            print(f" - Skipping sid {sid}")
            continue
        for (contour, color) in contour_names.items():
            if contour in sub_traces['iso_eccen']:
                tag = 'iso_eccen'
            else:
                tag = 'iso_angle'
            try:
                (x,y) = traces[rater][sid][h][tag][contour].points
            except Exception:
                print(f" - no points for {sid} / {contour}")
                continue
            # At this point we have (x,y) point-vectors for the
            # appropiate subject, hemisphere, rater, and contour name.
            ax.plot(x, y, '-', lw=lw, c=color, alpha=alpha)
        #print(f" - Finished {sid}/{h}")
    # Clean up axes a little:
    ax.axis('off')
```
